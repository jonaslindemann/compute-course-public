## What is Object-Oriented Programming?

Object-oriented programming (OOP) is a programming paradigm that organizes code around **objects** rather than functions. The basic concept is to combine data and methods into a single entity called an object.

### Key Concepts:
- **Objects** often represent nouns (things): point, circle, car, student, bank account
- **Objects interact** with each other by sending messages (calling methods)
- **Methods** are verbs that define what the object can do: move(), draw(), deposit(), withdraw()
- **Attributes** (properties) describe the object's state and relationships to other objects
- **Classes** are blueprints that define how objects should be structured and behave

### Benefits of OOP:
- **Code reusability**: Write once, use many times
- **Modularity**: Easier to maintain and debug
- **Abstraction**: Hide complex implementation details
- **Encapsulation**: Protect data and control access
- **Inheritance**: Build on existing code
- **Polymorphism**: Use objects of different types interchangeably

# Function-Based Programming vs Object-Oriented Programming

Before diving into OOP, let's first look at how we might solve problems using a **function-based** (procedural) approach. This will help us understand why OOP can be more effective for certain types of problems.

## Example: Managing Points in 2D Space

Let's say we want to represent and manipulate points in a 2D coordinate system. Using a function-based approach, we might represent a point as a list and create separate functions to work with points:

````python
# Function-based approach to working with 2D points
# A point is represented as a list [x, y]

def create_point(x, y):
    """Create a new point with given coordinates"""
    return [x, y]

def move_point(point, dx, dy):
    """Move a point by dx and dy"""
    point[0] += dx
    point[1] += dy

def zero_point(point):
    """Reset a point to origin (0, 0)"""
    point[0] = 0.0
    point[1] = 0.0

def set_point(point, x, y):
    """Set point coordinates to specific values"""
    point[0] = x
    point[1] = y

def print_point(point):
    """Print point coordinates in a readable format"""
    print(f"Point: x = {point[0]}, y = {point[1]}")

def distance_from_origin(point):
    """Calculate distance from point to origin"""
    return (point[0]**2 + point[1]**2)**0.5
````

````python
# Create a point and display it
p = create_point(0.5, 0.0)
print_point(p)
print(f"Distance from origin: {distance_from_origin(p):.2f}")
````

````python
# Move the point and display the new position
move_point(p, 3.0, 2.0)
print("After moving by (3.0, 2.0):")
print_point(p)
print(f"Distance from origin: {distance_from_origin(p):.2f}")
````

````python
# Set the point to specific coordinates
set_point(p, -2.0, -1.0)
print("After setting to (-2.0, -1.0):")
print_point(p)
print(f"Distance from origin: {distance_from_origin(p):.2f}")
````

````python
# Let's create multiple points to see potential issues
p1 = create_point(1.0, 2.0)
p2 = create_point(-1.0, 3.0)

print("Point 1:")
print_point(p1)
print("Point 2:")
print_point(p2)

# What if we accidentally pass wrong arguments?
# This could cause problems:
# move_point(3.0, p1, 2.0)  # Wrong order of arguments!
````

## Problems with Function-Based Approach

While the function-based approach works, it has several limitations:

1. **No data protection**: Anyone can modify the internal structure `point[0] = "invalid"`
2. **Scattered functionality**: Point-related functions are spread throughout the code
3. **Parameter confusion**: Easy to mix up parameter order: `move_point(dx, point, dy)`
4. **No type checking**: Nothing prevents passing wrong data types
5. **Difficult to extend**: Adding new point types (3D, polar coordinates) requires rewriting functions
6. **Global namespace pollution**: Many function names in the global scope

**Object-oriented programming solves these problems by bundling data and related functions together!**

# Encapsulation and Data Protection

**Encapsulation** is one of the core principles of OOP. It means:
- **Bundling** data and methods together in a class
- **Controlling access** to internal data
- **Hiding implementation details** from users of the class

## Why is Encapsulation Important?

1. **Data Integrity**: Prevent invalid values from being assigned
2. **Interface Stability**: Internal changes don't break external code
3. **Debugging**: Easier to track where data is modified
4. **Security**: Sensitive data can be protected

## The Problem with Direct Attribute Access

In our current Point class, anyone can directly modify the attributes:

```python
p = Point()
p.x = "invalid"  # This should not be allowed!
p.y = [1, 2, 3]  # Neither should this!
```

Let's see how Python provides encapsulation mechanisms.

## Private Attributes in Python

Python uses **name mangling** to create private attributes:
- Add double underscore prefix: `__attribute_name`
- Python automatically renames these internally
- Cannot be accessed directly from outside the class

**Benefits of private attributes:**
- Prevent accidental modification
- Force users to use intended interface methods
- Allow validation and error checking
- Enable future changes without breaking external code

Let's create a Point class with private attributes and constructor parameters:

````python
class Point:
    """Point class with private attributes and parameterized constructor"""

    def __init__(self, x=0.0, y=0.0):
        """
        Initialize point with given coordinates

        Args:
            x (float): x-coordinate, defaults to 0.0
            y (float): y-coordinate, defaults to 0.0
        """
        self.__x = x  # Private attribute (name mangled)
        self.__y = y  # Private attribute (name mangled)

    def get_coordinates(self):
        """Return tuple of (x, y) coordinates"""
        return (self.__x, self.__y)
````

We can now create **Point** instances and assign the instance attributes at the same time:

````python
# Create Point objects with initial values
p1 = Point()           # Uses default values (0.0, 0.0)
p2 = Point(1.0, 2.0)   # Custom coordinates
p3 = Point(x=5.0)      # Only x specified, y defaults to 0.0

print("Point coordinates:")
print(f"p1: {p1.get_coordinates()}")
print(f"p2: {p2.get_coordinates()}")
print(f"p3: {p3.get_coordinates()}")
````

We can no longer assign the instance attributes directly anymore. Something goes wrong:

````python
# Try to access attributes that don't exist (will cause AttributeError)
p = Point(1.0, 2.0)

try:
    print("Trying to access p.x:")
    print(p.x)
except AttributeError as e:
    print(f"Error: {e}")

try:
    print("Trying to access p.y:")
    print(p.y)
except AttributeError as e:
    print(f"Error: {e}")

print("\nBut we can access coordinates through the proper method:")
print(f"Coordinates: {p.get_coordinates()}")
````

The reason for this is that the x and y properties are now private. The internal attributes **__ x** cannot be reached either:

````python
# Private attributes cannot be accessed directly
p = Point(3.0, 4.0)

try:
    print("Trying to access private attribute __x:")
    print(p.__x)
except AttributeError as e:
    print(f"Error: {e}")

# However, Python name mangling means it's still accessible (but you shouldn't do this!)
print(f"\nActual private attribute name: {dir(p)}")
print("Private attributes are name-mangled to _ClassName__attribute")
print(f"p._Point__x = {p._Point__x}")  # Don't do this in real code!
print(f"p._Point__y = {p._Point__y}")  # This breaks encapsulation!
````

We have now protected our internal instance variables from a user of the object (encapsulation). To access these, we must now add methods to assign and return values. This is often done with get / set methods in object-oriented programming. We add the set methods to be able to assign the instance variables.

````python
class Point:
    def __init__(self, x=0.0, y=0.0):
        self.__x = x
        self.__y = y

    def x(self, value):
        self.__x = value

    def y(self, value):
        self.__y = value

    def set(self, x, y):
        self.__x = x
        self.__y = y
````

The class can now be created and used as in the code below:

````python
p = Point(12.0, 13.0)

p.set(2.0, 3.0)
p.x(42.0)
p.y(83.0)
````

To return values, we add the functions **x ()** and **y ()**.

````python
class Point:
    def __init__(self, x=0.0, y=0.0):
        self.__x = x
        self.__y = y

    def x(self):
        return self.__x

    def set_x(self, value):
        self.__x = value

    def y(self):
        return self.__y

    def set_y(self, value):
        self.__y = value

    def set(self, x, y):
        self.__x = x
        self.__y = y


````

Now we have full access to the internal instance variables in the class:

````python
p = Point(12.0, 13.0)

p.set(2.0, 3.0)
p.set_x(42.0)
p.set_y(83.0)

print(p.x())
print(p.y())
````

## Python Properties - The Pythonic Way

**Properties** provide a clean, Pythonic way to implement getters and setters:

### Without Properties (Verbose):
```python
p.set_x(5.0)     # Setting value
print(p.get_x()) # Getting value
```

### With Properties (Clean):
```python
p.x = 5.0        # Setting value (looks like direct access)
print(p.x)       # Getting value (looks like direct access)
```

### How Properties Work:
1. **Define getter/setter methods** for attribute access
2. **Use `property()` function** to create property objects
3. **Access attributes naturally** while maintaining encapsulation
4. **Add validation/logging** in setter methods

### Benefits:
- **Clean syntax**: Looks like direct attribute access
- **Backwards compatibility**: Can add properties later without breaking code
- **Data validation**: Control what values are allowed
- **Computed properties**: Calculate values on-the-fly

Let's see properties in action:

````python
class Point:
    """Point class demonstrating properties with validation"""

    def __init__(self, x=0.0, y=0.0):
        # Use the property setters for validation during initialization
        self.x = x  # This calls the setter!
        self.y = y  # This calls the setter!

    def get_x(self):
        """Getter method for x coordinate"""
        print("Getting x value")
        return self.__x

    def set_x(self, value):
        """Setter method for x coordinate with validation"""
        print(f"Setting x to {value}")
        if not isinstance(value, (int, float)):
            raise TypeError("x coordinate must be a number")
        self.__x = float(value)

    def get_y(self):
        """Getter method for y coordinate"""
        print("Getting y value")
        return self.__y

    def set_y(self, value):
        """Setter method for y coordinate with validation"""
        print(f"Setting y to {value}")
        if not isinstance(value, (int, float)):
            raise TypeError("y coordinate must be a number")
        if value > 100.0:  # Example constraint
            print("Warning: y value is very large, capping at 100.0")
            self.__y = 100.0
        else:
            self.__y = float(value)

    # Create property objects that connect attribute access to methods
    x = property(get_x, set_x, doc="X coordinate property")
    y = property(get_y, set_y, doc="Y coordinate property")
````

It is now much easier to access the instance variables:

````python
p = Point()

p.x = 42.0
p.y = 84.0

print(p.x)
print(p.y)
````

Properties give us the protection of get / set methods, but with the simplicity of assigning instance variables directly. It also gives us the opportunity to postpone the decision to add get / set methods and to use direct access to instance variables until you need the added protection.

## Python Properties Using Decorators - The Modern Way

**Decorators** provide an even cleaner way to create properties in Python. They are the preferred modern approach.

### `@property` decorator:
- Converts a method into a **read-only property**
- The method name becomes the property name
- Called automatically when accessing the attribute

### `@attribute.setter` decorator:
- Creates a **write-property** for an existing `@property`
- Must use the same method name as the property
- Called automatically when assigning to the attribute

### Advantages of Decorator Syntax:
- **Less code**: No need for separate `property()` calls
- **Cleaner**: Method definitions are grouped together
- **More readable**: Clear intent with decorators
- **Modern Python style**: Preferred in contemporary code

Here's our Point class rewritten using decorators:

````python
class Point:
    """Modern Point class using property decorators"""

    def __init__(self, x=0.0, y=0.0):
        """Initialize with validation through property setters"""
        self.x = x  # Calls the @x.setter method
        self.y = y  # Calls the @y.setter method

    @property
    def x(self):
        """Get x coordinate (read-only access)"""
        print("Reading x property")
        return self.__x

    @x.setter
    def x(self, value):
        """Set x coordinate with validation"""
        print(f"Setting x property to {value}")
        if not isinstance(value, (int, float)):
            raise TypeError(f"x must be a number, got {type(value)}")
        self.__x = float(value)

    @property
    def y(self):
        """Get y coordinate (read-only access)"""
        print("Reading y property")
        return self.__y

    @y.setter
    def y(self, value):
        """Set y coordinate with validation"""
        print(f"Setting y property to {value}")
        if not isinstance(value, (int, float)):
            raise TypeError(f"y must be a number, got {type(value)}")
        if value < -100 or value > 100:
            raise ValueError(f"y must be between -100 and 100, got {value}")
        self.__y = float(value)

    @property
    def distance_from_origin(self):
        """Computed property: distance from origin (read-only)"""
        return (self.__x**2 + self.__y**2)**0.5

    def __str__(self):
        """String representation of the point"""
        return f"Point({self.__x}, {self.__y})"
````

The class can be used in exactly the same way as before, but now we don't need the **property**-definitions at the end of the class.

````python
# Using properties with validation
p = Point()

print("=== Setting valid values ===")
p.x = 42.0
p.y = 84.0

print("\n=== Reading values ===")
print(f"x = {p.x}")
print(f"y = {p.y}")
print(f"Distance from origin = {p.distance_from_origin:.2f}")
print(f"Point: {p}")

print("\n=== Testing validation ===")
try:
    p.y = 150  # This will raise ValueError
except ValueError as e:
    print(f"Validation error: {e}")

try:
    p.x = "invalid"  # This will raise TypeError
except TypeError as e:
    print(f"Type error: {e}")
````

If you only want to implement a read-only property the @xxxx.setter decorator can be left out.

# Instance methods

*Main way of interacting with objects.* Due to the fact that all data is handled internally in the object, the methods can often be shortened.

A typical instance method in the **Point** class can be a method of moving the point a certain distance in the x and y directions.

````python
class Point:
    def __init__(self, x=0.0, y=0.0):
        self.__x = x
        self.__y = y

    @property
    def x(self):
        print("@property of x")
        return self.__x

    @x.setter
    def x(self, x):
        print("@setter for x")
        self.__x = x

    @property
    def y(self):
        print("@property of y")
        return self.__y

    @y.setter
    def y(self, y):
        print("@setter for y")
        self.__y = y

    def set(self, x, y):
        self.__x = x
        self.__y = y

    def move(self, dx, dy): # <---- New move method
        self.__x += dx
        self.__y += dy
````

We can now use **`.move(...)`** to move our points:

````python
p0 = Point()
p1 = Point()
p0.move(10.0, 20.0)
p1.move(-5.0, -5.0)

print(p0.x, p0.y)
print(p1.x, p1.y)
````

Another method that can be implemented is a method for copying the attributes from another instance of the same object **.copy_from()**:

````python
class Point:
    """Complete Point class with all methods"""

    def __init__(self, x=0.0, y=0.0):
        self.__x = x
        self.__y = y

    @property
    def x(self):
        return self.__x

    @x.setter
    def x(self, x):
        self.__x = x

    @property
    def y(self):
        return self.__y

    @y.setter
    def y(self, y):
        self.__y = y

    def set(self, x, y):
        """Set both coordinates at once"""
        self.__x = x
        self.__y = y

    def move(self, dx, dy):
        """Move point by dx and dy"""
        self.__x += dx
        self.__y += dy

    def copy_from(self, p):
        """Copy coordinates from another point"""
        self.__x = p.x
        self.__y = p.y
````

How this method is used is shown below:

````python
# Demonstrating the copy_from method
p0 = Point()
p1 = Point()

p0.move(10.0, 20.0)
p1.move(-5.0, -5.0)

print("Before copy_from:")
print(f"p0: x={p0.x}, y={p0.y}")  # Fixed: was (p0.x, p0.x)
print(f"p1: x={p1.x}, y={p1.y}")

# Copy p0's coordinates to p1
p1.copy_from(p0)

print("\nAfter p1.copy_from(p0):")
print(f"p0: x={p0.x}, y={p0.y}")
print(f"p1: x={p1.x}, y={p1.y}")
print("Notice: p1 now has the same coordinates as p0")
````

## Special instance methods

There are a variety of special instance methods for implementing extra functionality on objects **`__init__(...)`** is one of them
**`__str__(...)`** is used to convert an object to a printable form with **`print()`**

Using **`print()`** on an instance without the **`__str__()`** method produces the following printout.

````python
p = Point()
print(p)
````

It would be nicer if the item could print in a more presentable way:

````python
class Point:
    def __init__(self, x=0.0, y=0.0):
        self.__x = x
        self.__y = y

    @property
    def x(self):
        return self.__x

    @x.setter
    def x(self, x):
        self.__x = x

    @property
    def y(self):
        return self.__y

    @y.setter
    def y(self, y):
        self.__y = y

    def set(self, x, y):
        self.__x = x
        self.__y = y

    def move(self, dx, dy):
        self.__x += dx
        self.__y += dy

    def copy_from(self, p):
        self.__x = p.x
        self.__y = p.y

    def __str__(self): # <---- New __str__ method
        return "Point("+str(self.__x)+", "+str(self.__y)+")"
````

If we use the same code to print, we get instead:

````python
p = Point()

p.x = 42.0
p.y = 84.0

print(p)
````

## More special methods - operator overloading

`__init__()` and `__str__()` are only two of many *special methods*, also called *dunder methods* (for the **d**ouble **under**scores). Python calls them automatically when an object is used with the built-in operators and functions:

| Expression | Method called |
|---|---|
| `print(a)`, `str(a)` | `a.__str__()` |
| `repr(a)`, items in a printed list | `a.__repr__()` |
| `a + b` | `a.__add__(b)` |
| `a - b` | `a.__sub__(b)` |
| `a * 2.0` | `a.__mul__(2.0)` |
| `2.0 * a` | `a.__rmul__(2.0)` |
| `-a` | `a.__neg__()` |
| `abs(a)` | `a.__abs__()` |
| `a == b` | `a.__eq__(b)` |

By implementing these methods, our own classes can be used with the same syntax as the built-in number types. A typical example is a 2D vector class (`oop/1_basics/11_vector_dunder_methods.py`). We will use it again in the larger examples at the end of this notebook.

`Vector2` uses plain public attributes `x` and `y`. For a small value-like class such as this, that is a common and perfectly acceptable choice.

````python
import math

class Vector2:
    def __init__(self, x=0.0, y=0.0):
        self.x = x
        self.y = y

    def __repr__(self):
        return "Vector2(" + str(self.x) + ", " + str(self.y) + ")"

    def __str__(self):
        return "(" + str(self.x) + ", " + str(self.y) + ")"

    def __add__(self, other):
        return Vector2(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector2(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector2(self.x * scalar, self.y * scalar)

    def __rmul__(self, scalar):
        return self * scalar

    def __neg__(self):
        return Vector2(-self.x, -self.y)

    def __abs__(self):
        return math.sqrt(self.x**2 + self.y**2)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
````

`__str__()` gives a readable form for `print()`. `__repr__()` gives an unambiguous form, which looks like the code that creates the object. Lists use `__repr__()` when they show their items:

````python
a = Vector2(1.0, 2.0)
b = Vector2(3.0, 4.0)

print(a)                     # __str__
print(repr(a))               # __repr__
print([a, b])                # lists show their items with __repr__
````

The arithmetic operators now work on vectors:

````python
print(a + b)                 # a.__add__(b)
print(b - a)                 # b.__sub__(a)
print(a * 2.0)               # a.__mul__(2.0)
print(2.0 * a)               # float can't multiply a Vector2, so Python tries a.__rmul__(2.0)
print(-a)                    # a.__neg__()
print(abs(b))                # b.__abs__()
print(a == Vector2(1.0, 2.0))  # a.__eq__(...)
````

Look at `2.0 * a`. Python first tries `float.__mul__(2.0, a)`. The float type doesn't know about `Vector2`, so Python then tries the *reflected* method, `a.__rmul__(2.0)`.

With vector arithmetic, physics code reads like the equations. Here is an explicit Euler update of a position, $\mathbf{x} \leftarrow \mathbf{x} + \mathbf{v}\,\Delta t$:

````python
pos = Vector2(0.0, 0.0)
vel = Vector2(1.0, 0.5)
dt = 0.5

for i in range(5):
    pos += vel * dt          # no __iadd__, so Python uses pos = pos + vel * dt
    print(pos)
````

`Vector2` doesn't define `__iadd__()`, so `pos += vel * dt` runs as `pos = pos + vel * dt`. Each step creates a new `Vector2` object.

# Class Inheritance - Building on Existing Code

**Inheritance** allows you to create new classes based on existing ones, inheriting their attributes and methods while adding or modifying functionality.

## Key Concepts:

### Parent Class (Base/Super Class):
- The class being inherited from
- Provides common attributes and methods
- Example: `Point` class with x, y coordinates

### Child Class (Derived/Sub Class):
- The class that inherits from another class
- Gets all parent attributes and methods automatically
- Can add new features or override existing ones
- Example: `Circle` class inheriting from `Point`

## Inheritance Syntax:
```python
class ParentClass:
    # Parent implementation
    pass

class ChildClass(ParentClass):  # Inherit from ParentClass
    # Child implementation
    pass
```

## Benefits of Inheritance:
1. **Code Reuse**: Don't repeat common functionality
2. **Hierarchical Organization**: Natural relationships between classes
3. **Polymorphism**: Treat related objects uniformly
4. **Extensibility**: Add features without modifying existing code

## Real-world Example:
```
Vehicle (parent)
├── Car (inherits from Vehicle)
├── Motorcycle (inherits from Vehicle)
└── Truck (inherits from Vehicle)
```

Let's see inheritance in action with geometric shapes:

````python
class Circle(Point):  # Circle inherits from Point
    """
    Circle class that inherits from Point
    A circle is a point with a radius
    """

    def __init__(self, x=0.0, y=0.0, r=1.0):
        """
        Initialize a circle with center coordinates and radius

        Args:
            x (float): x-coordinate of center
            y (float): y-coordinate of center
            r (float): radius of the circle
        """
        # Call the parent class constructor to initialize x and y
        super().__init__(x, y)
        self.r = r  # Add radius property using the setter

    @property
    def r(self):
        """Get radius of the circle"""
        return self.__r

    @r.setter
    def r(self, value):
        """Set radius with validation"""
        if not isinstance(value, (int, float)):
            raise TypeError("Radius must be a number")
        if value <= 0:
            raise ValueError("Radius must be positive")
        self.__r = float(value)

    @property
    def area(self):
        """Calculate area of the circle (computed property)"""
        import math
        return math.pi * self.__r ** 2

    @property
    def circumference(self):
        """Calculate circumference of the circle (computed property)"""
        import math
        return 2 * math.pi * self.__r

    def copy_from(self, other_circle):
        """
        Copy data from another circle
        Overrides parent method to also copy radius
        """
        super().copy_from(other_circle)  # Copy x, y coordinates
        self.r = other_circle.r           # Copy radius

    def __str__(self):
        """String representation of the circle"""
        return f"Circle(center=({self.x}, {self.y}), radius={self.__r})"
````

## Using Inherited Classes

The `Circle` class now has:
- **All Point methods**: x, y properties, move(), distance_from_origin
- **Additional Circle methods**: r property, area, circumference
- **Overridden methods**: copy_from(), __str__()

Let's see inheritance in action:

````python
# Create a Point and a Circle
p = Point(1.0, 2.0)
c = Circle(2.0, 4.0, 3.0)

print("=== Initial objects ===")
print(f"Point: {p}")
print(f"Circle: {c}")

print("\n=== Circle-specific properties ===")
print(f"Circle radius: {c.r}")
print(f"Circle area: {c.area:.2f}")
print(f"Circle circumference: {c.circumference:.2f}")

print("\n=== Inherited functionality ===")
print(f"Circle center coordinates: x={c.x}, y={c.y}")
print(f"Point coordinates: x={p.x}, y={p.y}")

# Both Point and Circle have move() method (inherited)
print("\n=== Moving objects ===")
print("Before move:")
print(f"  Point: {p}")
print(f"  Circle: {c}")

p.move(1.0, 1.0)
c.move(1.0, 1.0)

print("After moving both objects by (1, 1):")
print(f"  Point: {p}")
print(f"  Circle: {c}")

print("\n=== Modifying circle properties ===")
c.r = 5.0
c.x = 0.0
c.y = 0.0
print(f"Modified circle: {c}")
print(f"New area: {c.area:.2f}")
````

# Composition - Objects Containing Other Objects

**Composition** is when objects contain references to other objects as part of their structure. Instead of inheriting behavior, objects are built by combining simpler objects.

## Composition vs Inheritance:

### Inheritance ("is-a" relationship):
- `Circle` **is a** `Point` (with added radius)
- Child class extends parent class
- Shares the same interface

### Composition ("has-a" relationship):
- `Line` **has** two `Point` objects (start and end points)
- `Car` **has** an `Engine` object
- `Rectangle` **has** four `Point` objects (corners)

## When to Use Composition:
- When objects are naturally made of other objects
- When you want more flexibility than inheritance provides
- When the relationship is "has-a" rather than "is-a"
- When you need to change behavior at runtime

## Benefits:
1. **Flexibility**: Can change components at runtime
2. **Loose Coupling**: Objects can be developed independently  
3. **Code Reuse**: Compose different combinations of objects
4. **Clear Relationships**: Models real-world "has-a" relationships

## Example: Line Class
A line connects two points, so it **contains** two Point objects:

````python
import math

class Line:
    """
    Line class demonstrating composition
    A line is composed of two Point objects
    """

    def __init__(self, p0=None, p1=None):
        """
        Initialize line with two points

        Args:
            p0: First point (start). If None, creates Point(0,0)
            p1: Second point (end). If None, creates Point(1,1)
        """
        # Composition: Line contains Point objects
        self.__p0 = p0 if p0 is not None else Point()
        self.__p1 = p1 if p1 is not None else Point(1.0, 1.0)

    @property
    def p0(self):
        """Get the first point of the line"""
        return self.__p0

    @p0.setter
    def p0(self, point):
        """Set the first point of the line"""
        if not isinstance(point, Point):
            raise TypeError("p0 must be a Point object")
        self.__p0 = point

    @property
    def p1(self):
        """Get the second point of the line"""
        return self.__p1

    @p1.setter
    def p1(self, point):
        """Set the second point of the line"""
        if not isinstance(point, Point):
            raise TypeError("p1 must be a Point object")
        self.__p1 = point

    @property
    def length(self):
        """Calculate length of the line (computed property)"""
        dx = self.__p1.x - self.__p0.x
        dy = self.__p1.y - self.__p0.y
        return math.sqrt(dx*dx + dy*dy)

    @property
    def midpoint(self):
        """Calculate midpoint of the line (computed property)"""
        mid_x = (self.__p0.x + self.__p1.x) / 2
        mid_y = (self.__p0.y + self.__p1.y) / 2
        return Point(mid_x, mid_y)

    def move(self, dx, dy):
        """Move the entire line by displacing both points"""
        self.__p0.move(dx, dy)
        self.__p1.move(dx, dy)  # Fixed: was (dy, dy)

    def __str__(self):
        """String representation of the line"""
        return f"Line from {self.__p0} to {self.__p1} (length: {self.length:.2f})"
````

Note that the **Line** class does not inherit from **Point**. It only refers to instances of type **Point**.

````python
# Create a line using default points
line = Line()
print(f"Default line: {line}")

# Modify points through properties
line.p0.x = 0.0
line.p0.y = 2.0
line.p1.x = 3.0
line.p1.y = 4.0

print(f"Modified line: {line}")
print(f"Line length: {line.length:.2f}")
print(f"Line midpoint: {line.midpoint}")

# Create line with custom points
start_point = Point(-1.0, -1.0)
end_point = Point(2.0, 3.0)
custom_line = Line(start_point, end_point)

print(f"\nCustom line: {custom_line}")

# Move the entire line
custom_line.move(1.0, 1.0)
print(f"After moving by (1,1): {custom_line}")
````

In this example, the **Point** instances are created by the **Line** and the setters check that only **Point** objects are assigned. The **x** and **y** attributes of the contained **Point** objects can still be modified through the **p0** and **p1** properties, as in `line.p0.x = 0.0`.

A simpler version, closer to `oop/1_basics/09_composition_line.py`, creates two default points and allows the instance properties **p0** and **p1** to be assigned without checks:

````python
class Line:
    def __init__(self):
        self.__p0 = Point()
        self.__p1 = Point()

    @property
    def p0(self):
        return self.__p0

    @p0.setter
    def p0(self, p):
        self.__p0 = p

    @property
    def p1(self):
        return self.__p1

    @p1.setter
    def p1(self, p):
        self.__p1 = p

    def __str__(self):
        return "Line from: " + str(self.__p0) + " to " + str(self.__p1)
````

The class can now be used as follows:

````python
line = Line()
line.p0.x = 0.0
line.p0.y = 2.0
line.p1.x = 3.0
line.p1.y = 4.0

print(line)

p3 = Point(5.0, 5.0)
p4 = Point(10.0, 10.0)
line.p0 = p3
line.p1 = p4

print(line)
````

A method for calculating the length can easily be added:

````python
import math

class Line:
    def __init__(self):
        self.__p0 = Point()
        self.__p1 = Point()

    @property
    def p0(self):
        return self.__p0

    @p0.setter
    def p0(self, p):
        self.__p0 = p

    @property
    def p1(self):
        return self.__p1

    @p1.setter
    def p1(self, p):
        self.__p1 = p

    @property
    def length(self):
        return math.sqrt(math.pow(self.p1.x - self.p0.x, 2) +
            math.pow(self.p1.y - self.p0.y, 2))

    def __str__(self):
        return "Line from: " + str(self.__p0) + " to " + str(self.__p1)
````

Examples of use:

````python
line = Line()
line.p0.x = 0.0
line.p0.y = 2.0
line.p1.x = 3.0
line.p1.y = 4.0

print(line.length)
````

# Polymorphism - One Interface, Many Forms

**Polymorphism** means "many forms" - it allows objects of different types to be treated uniformly through a common interface.

## Key Concepts:

### Dynamic Method Resolution:
- The **correct method is called** based on the actual object type
- Happens at **runtime**, not compile time
- If a method isn't found in the current class, Python looks in parent classes

### Duck Typing (Python's Approach):
> *"If it looks like a duck and quacks like a duck, it's probably a duck"*

- Python doesn't care about the object's type
- Only cares that the object has the required methods
- If an object has a `move()` method, it can be moved

## Benefits:
1. **Code Flexibility**: Work with objects without knowing their exact type
2. **Extensibility**: Add new types without changing existing code  
3. **Clean Interfaces**: Single way to interact with related objects
4. **Runtime Behavior**: Choose behavior based on actual object type

## Real-world Examples:
- All shapes can be `draw()`-n, regardless of whether they're circles, rectangles, or triangles
- All files can be `read()` from, whether they're text files, images, or videos
- All vehicles can `start()`, whether they're cars, motorcycles, or boats

Let's see polymorphism in action:

````python
# Polymorphism example: Different objects in the same list
shapes = []

# Add different types of objects to the list
shapes.append(Point(0.0, 1.0))           # Point object
shapes.append(Circle(2.0, 1.0, 3.0))     # Circle object
shapes.append(Line())                     # Line object
shapes.append(42.0)                       # float object
shapes.append(42)                         # int object
shapes.append("Hello")                    # string object

print("=== Polymorphism in action ===")
print("All objects in the list, regardless of type:")

# Polymorphic behavior: print() works on all objects
for i, shape in enumerate(shapes):
    print(f"{i}: {shape}")
    print(f"    Type: {type(shape).__name__}")

print("\n=== Every object has a __str__ method ===")
# This is why print() works - every object has __str__()
for shape in shapes:
    print(f"Calling __str__(): {shape.__str__()}")
````

What does print (42.0) really mean?

````python
print(42.0.__str__())
````

````python
# Polymorphism with geometric shapes that share common methods
geometric_shapes = []

geometric_shapes.append(Point(0.0, 1.0))
geometric_shapes.append(Circle(2.0, 1.0, 3.0))

print("=== Polymorphic method calls ===")
print("Both Point and Circle objects have move() method:")

for shape in geometric_shapes:
    print(f"Before move: {shape}")
    shape.move(5, 5)  # Polymorphic call - works on both types
    print(f"After move(5,5): {shape}")
    print()

print("=== Different behavior, same method name ===")
# Different classes can implement the same method differently
shapes_with_areas = [
    Circle(0, 0, 3),     # Circle has area = π * r²
    # We could add Rectangle, Triangle, etc. here
]

for shape in shapes_with_areas:
    if hasattr(shape, 'area'):  # Check if object has area property
        print(f"{shape}")
        print(f"  Area: {shape.area:.2f}")
    if hasattr(shape, 'circumference'):  # Check if object has circumference
        print(f"  Circumference: {shape.circumference:.2f}")
    print()
````

### The class hierarchy


`DrawableBase.draw()` sets the colours and the line width, then calls `on_draw()` to do the actual drawing. The base class `on_draw()` does nothing. Each subclass overrides it with its own shape (from `03_particle_inheritance.py` and `04_round_and_square_particles.py`):

```python
class DrawableBase:
    def __init__(self):
        self.stroke_color = [0, 0, 0]
        self.stroke_alpha = 255
        self.fill_color = [255, 255, 255]
        self.fill_alpha = 255
        self.stroke_width = 1

    def draw(self):
        sketch.stroke(self.stroke_color[0], self.stroke_color[1], self.stroke_color[2], self.stroke_alpha)
        sketch.fill(self.fill_color[0], self.fill_color[1], self.fill_color[2], self.fill_alpha)
        sketch.stroke_weight(self.stroke_width)

        self.on_draw()

    def on_draw(self):
        pass

class Particle(DrawableBase):
    def __init__(self, x=0.0, y=0.0):
        super().__init__()
        self.x = x
        self.y = y
        self.vx = 0.0
        self.vy = 0.
        self.stroke_color = [255, 0, 0]

        self.stroke_width = 2

    def update(self, dt):
        self.x += self.vx*dt
        self.y += self.vy*dt

    def on_draw(self):
        sketch.point(self.x, self.y)

class RoundParticle(Particle):
    def __init__(self, x=0.0, y=0.0, r=50.0):
        super().__init__(x, y)
        self.r = r

    def on_draw(self):
        sketch.ellipse(self.x, self.y, self.r*2, self.r*2)

class SquareParticle(Particle):
    def __init__(self, x=0.0, y=0.0, w=50.0, h=50.0):
        super().__init__(x, y)
        self.w = w
        self.h = h
        self.r = w/2

    def on_draw(self):
        sketch.rect(self.x, self.y, self.w, self.h)
```

This is a common design pattern, the *template method*. The base class defines the steps, and the subclasses fill in the parts that differ. A new particle type only needs a new subclass with an `on_draw()` method.

`BoxBoundary.check()` uses `p.r` to bounce a particle off the walls. That's why `SquareParticle` also sets `self.r = w/2`. `BoxBoundary` doesn't check the type of the particle; it only needs the attributes `x`, `y`, `vx`, `vy` and `r`.

### Vectors make the physics simpler

```python
    self.y += self.vy*dt

    self.pos += self.vel * dt
```

## Example 2: A predator-prey ecosystem ([`oop/7_ecosystem`](https://github.com/jonaslindemann/compute-course-public/tree/master/oop/7_ecosystem))

A small world with grass, rabbits and foxes:

- **Grass** is eaten by rabbits and regrows after a while.
- **Rabbits** eat grass and flee from foxes.
- **Foxes** hunt rabbits when they are hungry.
- All animals use energy as they live. With enough energy they reproduce, and with none left they die.

These simple rules produce populations that rise and fall in cycles, as in the classic Lotka-Volterra equations. Here, though, every animal is an individual object. This is called an *agent-based* model.

### Design

```
Agent              pos, vel, energy, update(), behave(), move(), reproduce(), draw()
├── Grass          regrows, overrides update() and draw()
├── Rabbit         overrides behave(): flee, eat grass, wander
└── Fox            overrides behave(): hunt; overrides draw()
World              list of agents, step(), nearest(), count()
Vector2            2D vector with operator overloading
```

### Vector2 with a class method

The ecosystem uses an extended `Vector2` with two new features:

- `from_angle()` is a **class method** (`@classmethod`). It receives the class `cls` instead of an instance and works as an alternative constructor: `Vector2.from_angle(angle, length)`.
- `angle` is a read-only property, and `normalized()` returns a vector with length 1 in the same direction.

````python
import math, random

class Vector2:
    def __init__(self, x=0.0, y=0.0):
        self.x = x
        self.y = y

    @classmethod
    def from_angle(cls, angle, length=1.0):
        return cls(math.cos(angle) * length, math.sin(angle) * length)

    def __repr__(self):
        return "Vector2(" + str(self.x) + ", " + str(self.y) + ")"

    def __str__(self):
        return "(" + str(self.x) + ", " + str(self.y) + ")"

    def __add__(self, other):
        return Vector2(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Vector2(self.x - other.x, self.y - other.y)

    def __mul__(self, scalar):
        return Vector2(self.x * scalar, self.y * scalar)

    def __rmul__(self, scalar):
        return self * scalar

    def __neg__(self):
        return Vector2(-self.x, -self.y)

    def __abs__(self):
        return math.sqrt(self.x**2 + self.y**2)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    @property
    def angle(self):
        return math.atan2(self.y, self.x)

    def normalized(self):
        length = abs(self)
        if length == 0.0:
            return Vector2()
        return self * (1.0 / length)
````

### The `Agent` base class

All agents share the same life cycle in `update()`. An agent decides what to do (`behave()`), moves, uses energy, and then dies or reproduces. Three things to notice:

- **Class attributes** like `color`, `speed` and `metabolism` are defined in the class body, not in `__init__()`. All objects of the class share them, and a subclass overrides them by defining new values. A rabbit gets `self.speed == 40.0` and a fox gets `43.0` without any extra code.
- `behave()` does nothing here. It is a *hook* for the subclasses to override.
- `reproduce()` creates the offspring with `type(self)(...)`. `type(self)` is the class of the actual object, so rabbits create rabbits and foxes create foxes. One implementation in the base class serves every subclass.

````python
class Agent:
    """Base class for everything that lives in the world."""

    # Class attributes: shared by all agents of a class, overridden in subclasses.
    color = (255, 255, 255)
    radius = 4.0
    speed = 0.0
    vision = 0.0
    metabolism = 0.0        # energy used per second
    birth_energy = None     # energy needed to reproduce, None = never reproduces

    def __init__(self, pos, energy=10.0):
        self.pos = pos
        self.vel = Vector2()
        self.heading = random.uniform(0.0, 2.0 * math.pi)
        self.energy = energy
        self.alive = True

    def update(self, world, dt):
        self.behave(world, dt)
        self.move(world, dt)

        self.energy -= self.metabolism * dt

        if self.energy <= 0.0:
            self.die()
        elif self.birth_energy is not None and self.energy > self.birth_energy:
            self.reproduce(world)

    def behave(self, world, dt):
        """Decide where to go. Overridden by the animals."""
        pass

    def move(self, world, dt):
        self.pos += self.vel * dt
        world.keep_inside(self)

    def wander(self, dt):
        self.heading += random.uniform(-3.0, 3.0) * dt
        self.vel = Vector2.from_angle(self.heading, self.speed)

    def move_towards(self, target):
        self.vel = (target.pos - self.pos).normalized() * self.speed
        self.heading = self.vel.angle

    def move_away_from(self, target):
        self.move_towards(target)
        self.vel = -self.vel
        self.heading = self.vel.angle

    def touches(self, other):
        return abs(other.pos - self.pos) < self.radius + other.radius

    def reproduce(self, world):
        self.energy /= 2.0

        # type(self) is the class of this object, so rabbits create
        # rabbits and foxes create foxes, without overriding this method.

        offset = Vector2(random.uniform(-5.0, 5.0), random.uniform(-5.0, 5.0))
        world.add(type(self)(self.pos + offset, self.energy))

    def die(self):
        self.alive = False

    def draw(self, canvas):
        canvas.circle(self.pos, self.radius, self.color)
````

### Grass, rabbits and foxes

The subclasses only contain what is different: their class attribute values and the methods they override.

- `Grass` doesn't move, so it replaces `update()` completely.
- `Rabbit` and `Fox` only override `behave()`. They keep the rest of the life cycle from `Agent`.
- `grown` and `hungry` are computed read-only properties.

````python
class Grass(Agent):
    color = (60, 170, 60)
    eaten_color = (45, 65, 40)
    radius = 3.0
    regrow_time = 10.0

    def __init__(self, pos, energy=0.0):
        super().__init__(pos, energy)
        self.regrow_timer = 0.0

    @property
    def grown(self):
        return self.regrow_timer <= 0.0

    def update(self, world, dt):
        # Grass doesn't move, use energy or reproduce. It only regrows.
        if self.regrow_timer > 0.0:
            self.regrow_timer -= dt

    def eaten(self):
        self.regrow_timer = self.regrow_time

    def draw(self, canvas):
        if self.grown:
            canvas.square(self.pos, 2.0 * self.radius, self.color)
        else:
            canvas.square(self.pos, 2.0 * self.radius, self.eaten_color)
````

````python
class Rabbit(Agent):
    color = (235, 235, 235)
    radius = 4.0
    speed = 40.0
    vision = 50.0
    metabolism = 1.0
    birth_energy = 20.0
    food_energy = 4.0

    def behave(self, world, dt):
        fox = world.nearest(self.pos, Fox, self.vision)

        if fox is not None:
            self.move_away_from(fox)
            return

        grass = world.nearest(self.pos, Grass, self.vision, lambda g: g.grown)

        if grass is None:
            self.wander(dt)
        elif self.touches(grass):
            grass.eaten()
            self.energy += self.food_energy
        else:
            self.move_towards(grass)
````

````python
class Fox(Agent):
    color = (235, 120, 40)
    radius = 6.0
    speed = 43.0
    vision = 70.0
    metabolism = 0.7
    birth_energy = 30.0
    food_energy = 15.0
    hungry_energy = 25.0    # foxes only hunt when their energy is below this

    @property
    def hungry(self):
        return self.energy < self.hungry_energy

    def behave(self, world, dt):
        rabbit = None

        if self.hungry:
            rabbit = world.nearest(self.pos, Rabbit, self.vision)

        if rabbit is None:
            self.wander(dt)
        elif self.touches(rabbit):
            rabbit.die()
            self.energy += self.food_energy
        else:
            self.move_towards(rabbit)

    def draw(self, canvas):
        canvas.triangle(self.pos, 2.0 * self.radius, self.heading, self.color)
````

`Rabbit` refers to `Fox`, and `Fox` refers to `Rabbit`. This works because Python looks the names up when `behave()` is called, and by then both classes are defined.

### The `World` - composition and polymorphism

`World` *has* a list of agents (composition). Its `step()` method is where polymorphism does its work:

```python
for agent in self.agents:
    if agent.alive:
        agent.update(self, dt)
```

Depending on the type of each object, this one call runs either `Grass.update()` or `Agent.update()`. `Agent.update()` in turn calls `Rabbit.behave()` or `Fox.behave()`. Also note:

- `nearest()` and `count()` take a **class as an argument** and use `isinstance()` to find agents of that type.
- New agents are collected in the private list `__newborn` and added after the loop. A list shouldn't be changed while you loop over it.

````python
class World:
    def __init__(self, width=800.0, height=600.0):
        self.width = width
        self.height = height
        self.time = 0.0
        self.agents = []

        # Migrants arriving from outside, per second. Without them a species that
        # dies out is gone forever, which happens easily in a small world.
        self.migration = {Rabbit: 0.1, Fox: 0.02}

        self.__newborn = []
        self.__type_cache = {}

    def random_pos(self):
        return Vector2(random.uniform(0.0, self.width), random.uniform(0.0, self.height))

    def populate(self, n_grass=250, n_rabbits=40, n_foxes=6):
        for i in range(n_grass):
            self.agents.append(Grass(self.random_pos()))
        for i in range(n_rabbits):
            self.agents.append(Rabbit(self.random_pos(), random.uniform(5.0, 15.0)))
        for i in range(n_foxes):
            self.agents.append(Fox(self.random_pos(), random.uniform(15.0, 30.0)))

    def random_edge_pos(self):
        if random.random() < 0.5:
            return Vector2(random.choice([0.0, self.width]), random.uniform(0.0, self.height))
        else:
            return Vector2(random.uniform(0.0, self.width), random.choice([0.0, self.height]))

    def add(self, agent):
        # Newborns join after the current step, so the list isn't changed while we loop over it.
        self.__newborn.append(agent)

    def step(self, dt):
        self.__type_cache = {}

        for cls, rate in self.migration.items():
            if random.random() < rate * dt:
                self.add(cls(self.random_edge_pos()))

        for agent in self.agents:
            if agent.alive:
                agent.update(self, dt)

        self.agents = [agent for agent in self.agents if agent.alive] + self.__newborn
        self.__newborn = []
        self.time += dt

    def of_type(self, cls):
        """All agents that are instances of cls, including subclasses of cls."""
        if cls not in self.__type_cache:
            self.__type_cache[cls] = [agent for agent in self.agents if isinstance(agent, cls)]
        return self.__type_cache[cls]

    def count(self, cls):
        return len([agent for agent in self.agents if isinstance(agent, cls) and agent.alive])

    def nearest(self, pos, cls, max_distance, condition=None):
        nearest_agent = None
        nearest_dist2 = max_distance**2

        for agent in self.of_type(cls):
            if not agent.alive or (condition is not None and not condition(agent)):
                continue

            dx = agent.pos.x - pos.x
            dy = agent.pos.y - pos.y
            dist2 = dx*dx + dy*dy

            if dist2 < nearest_dist2:
                nearest_agent = agent
                nearest_dist2 = dist2

        return nearest_agent

    def keep_inside(self, agent):
        if agent.pos.x < agent.radius:
            agent.pos.x = agent.radius
            agent.heading = math.pi - agent.heading
        elif agent.pos.x > self.width - agent.radius:
            agent.pos.x = self.width - agent.radius
            agent.heading = math.pi - agent.heading
        if agent.pos.y < agent.radius:
            agent.pos.y = agent.radius
            agent.heading = -agent.heading
        elif agent.pos.y > self.height - agent.radius:
            agent.pos.y = self.height - agent.radius
            agent.heading = -agent.heading
````

### Running the simulation

`World` doesn't know how it will be presented. Here we run it for 600 simulated seconds and record the populations, as in `01_population_plot.py`. The cell takes about 20 seconds to run.

````python
import matplotlib.pyplot as plt

random.seed(0)    # remove this line to get a different ecosystem every run

DT = 0.1          # seconds per simulation step
SIM_TIME = 600.0  # seconds to simulate

world = World()
world.populate()

times = []
rabbits = []
foxes = []

step = 0

while world.time < SIM_TIME:
    world.step(DT)
    step += 1

    if step % 5 == 0:
        times.append(world.time)
        rabbits.append(world.count(Rabbit))
        foxes.append(world.count(Fox))

    if step % 1000 == 0:
        print("t =", round(world.time), "s, rabbits:", rabbits[-1], "foxes:", foxes[-1])

fig, (ax_time, ax_phase) = plt.subplots(1, 2, figsize=(12, 4.5))

ax_time.plot(times, rabbits, color="tab:blue", label="Rabbits")
ax_time.plot(times, foxes, color="tab:orange", label="Foxes")
ax_time.set_xlabel("Time [s]")
ax_time.set_ylabel("Population")
ax_time.set_title("Populations over time")
ax_time.legend()

ax_phase.plot(rabbits, foxes, color="tab:gray", linewidth=0.8)
ax_phase.set_xlabel("Rabbits")
ax_phase.set_ylabel("Foxes")
ax_phase.set_title("Rabbits vs. foxes")

fig.tight_layout()
plt.show()
````

### Duck typing: drawing the world

Every agent draws itself with `agent.draw(canvas)` by calling `canvas.circle()`, `canvas.square()` or `canvas.triangle()`. The agents don't know what the canvas is. **Any object with these three methods works.** This is *duck typing*. Here is a canvas that draws with matplotlib:

````python
from matplotlib.patches import Circle as CirclePatch, Rectangle, Polygon

class MatplotlibCanvas:
    """Translates the drawing calls of the agents into matplotlib patches."""

    def __init__(self, ax):
        self.ax = ax

    def to_rgb(self, color):
        return [c / 255 for c in color]

    def circle(self, pos, radius, color):
        self.ax.add_patch(CirclePatch((pos.x, pos.y), radius, color=self.to_rgb(color)))

    def square(self, pos, size, color):
        self.ax.add_patch(Rectangle((pos.x - size / 2, pos.y - size / 2), size, size, color=self.to_rgb(color)))

    def triangle(self, pos, size, angle, color):
        tip = pos + Vector2.from_angle(angle, size)
        left = pos + Vector2.from_angle(angle + 2.5, size * 0.8)
        right = pos + Vector2.from_angle(angle - 2.5, size * 0.8)
        self.ax.add_patch(Polygon([(tip.x, tip.y), (left.x, left.y), (right.x, right.y)], color=self.to_rgb(color)))
````

````python
random.seed(1)

world = World()
world.populate()

for i in range(500):     # 50 simulated seconds
    world.step(0.1)

fig, ax = plt.subplots(figsize=(8, 6))
ax.set_facecolor((30 / 255, 40 / 255, 30 / 255))

canvas = MatplotlibCanvas(ax)

for agent in world.agents:
    agent.draw(canvas)          # polymorphic call, each agent draws itself

ax.set_xlim(0, world.width)
ax.set_ylim(world.height, 0)    # y axis pointing down, as on the screen
ax.set_aspect("equal")
ax.set_title(f"t = {world.time:.0f} s, rabbits: {world.count(Rabbit)}, foxes: {world.count(Fox)}")
plt.show()
````

## Example 3: Analysis of a truss bridge ([`oop/8_structures`](https://github.com/jonaslindemann/compute-course-public/tree/master/oop/8_structures))

The last example is an engineering application. It is a small program for linear static analysis of 2D trusses, which are structures made of bars connected by pin joints. It works like the finite element programs used in structural mechanics. We use it to analyse a Pratt truss bridge resting on two rubber bearings.

### Design

```
Node                 x, y, displacements ux, uy, dofs
AxialElement         abstract: stiffness_matrix(), compute_force()
├── Bar              axial_stiffness() = E·A/L
└── Spring           axial_stiffness() = k
Support              abstract: fixed_dofs()
├── PinSupport       fixed in x and y
└── RollerSupport    fixed in one direction
Structure            nodes, elements, supports, loads, solve()
```

### Nodes

````python
import numpy as np

class Node:
    def __init__(self, x, y):
        self.x = x
        self.y = y
        self.index = None    # set by Structure.add_node()
        self.ux = 0.0        # displacements, set by Structure.solve()
        self.uy = 0.0

    @property
    def dofs(self):
        """The global degrees of freedom of this node: [x, y]."""
        return [2*self.index, 2*self.index + 1]

    def __repr__(self):
        return "Node(" + str(self.x) + ", " + str(self.y) + ")"
````

`dofs` is a computed property. It gives the global *degrees of freedom* of the node, which are the positions of its x and y displacements in the global system of equations.

### An abstract base class for elements

Bars and springs both carry force only along their axis. Their stiffness matrices have the same form:

$$\mathbf{K}_e = k\,\mathbf{b}\,\mathbf{b}^T, \qquad \mathbf{b} = \begin{bmatrix}-c & -s & c & s\end{bmatrix}^T$$

Here $c$ and $s$ are the cosine and sine of the element angle. The only difference is the axial stiffness $k$. It is $EA/L$ for a bar and a given constant for a spring. All the shared code therefore lives in the base class `AxialElement`, and the subclasses only implement `axial_stiffness()`.

`AxialElement` doesn't know its own stiffness. Its `axial_stiffness()` raises `NotImplementedError`, which makes it an *abstract* base class: a class meant to be inherited from, not used directly.

````python
class AxialElement:
    """
    Base class for elements that only carry force along the line between their
    two nodes. Subclasses only have to say how stiff they are.
    """

    def __init__(self, node0, node1):
        self.node0 = node0
        self.node1 = node1
        self.force = 0.0     # axial force after solve(), positive = tension

    @property
    def length(self):
        return np.hypot(self.node1.x - self.node0.x, self.node1.y - self.node0.y)

    @property
    def dofs(self):
        return self.node0.dofs + self.node1.dofs

    def axial_stiffness(self):
        raise NotImplementedError("Subclasses of AxialElement must implement axial_stiffness()")

    def elongation_vector(self):
        """Elongation of the element = elongation_vector() @ [ux0, uy0, ux1, uy1]."""
        c = (self.node1.x - self.node0.x) / self.length
        s = (self.node1.y - self.node0.y) / self.length
        return np.array([-c, -s, c, s])

    def stiffness_matrix(self):
        b = self.elongation_vector()
        return self.axial_stiffness() * np.outer(b, b)

    def compute_force(self, u):
        self.force = self.axial_stiffness() * (self.elongation_vector() @ u[self.dofs])
````

````python
class Bar(AxialElement):
    """A bar with Young's modulus E [Pa] and cross-section area A [m2]."""

    def __init__(self, node0, node1, E=210e9, A=2e-3):
        super().__init__(node0, node1)
        self.E = E
        self.A = A

    def axial_stiffness(self):
        return self.E * self.A / self.length

    @property
    def stress(self):
        return self.force / self.A

    def __repr__(self):
        return "Bar(" + str(self.node0.index) + " -> " + str(self.node1.index) + ")"
````

````python
class Spring(AxialElement):
    """A spring with stiffness k [N/m], independent of its length."""

    def __init__(self, node0, node1, k):
        super().__init__(node0, node1)
        self.k = k

    def axial_stiffness(self):
        return self.k

    def __repr__(self):
        return "Spring(" + str(self.node0.index) + " -> " + str(self.node1.index) + ")"
````

If you try to use the base class directly, the error message says exactly what is missing:

````python
n0 = Node(0.0, 0.0)
n0.index = 0
n1 = Node(2.0, 0.0)
n1.index = 1

try:
    AxialElement(n0, n1).stiffness_matrix()
except NotImplementedError as e:
    print("Error:", e)

print(Bar(n0, n1).stiffness_matrix())
````

The standard library module `abc` offers a stricter version of this. A method marked `@abstractmethod` stops the base class from being instantiated at all. Raising `NotImplementedError` is a simpler way to get almost the same effect.

### Supports

The supports use the same pattern. Each subclass returns the degrees of freedom it fixes:

````python
class Support:
    """Base class for supports. Subclasses decide which directions are fixed."""

    def __init__(self, node):
        self.node = node

    def fixed_dofs(self):
        raise NotImplementedError("Subclasses of Support must implement fixed_dofs()")

class PinSupport(Support):
    """Fixed in both x and y."""

    def fixed_dofs(self):
        return self.node.dofs

class RollerSupport(Support):
    """Fixed in one direction, free to roll in the other ("x" or "y")."""

    def __init__(self, node, free="x"):
        super().__init__(node)
        self.free = free

    def fixed_dofs(self):
        if self.free == "x":
            return [self.node.dofs[1]]
        else:
            return [self.node.dofs[0]]
````

### The `Structure` class

A `Structure` is composed of nodes, elements, supports and loads. The `add_...()` methods create the objects and keep track of them, so the user never has to number the nodes by hand.

The heart of the program is `solve()`. It assembles the global stiffness matrix by asking every element for its own stiffness matrix:

```python
for element in self.elements:
    K[np.ix_(element.dofs, element.dofs)] += element.stiffness_matrix()
```

This loop is polymorphic. It works for bars, springs or any `AxialElement` subclass added later, without any change. `solve()` then solves the system $\mathbf{K}\mathbf{u} = \mathbf{f}$ for the free degrees of freedom and stores the results back in the node and element objects.

````python
class Structure:
    def __init__(self):
        self.nodes = []
        self.elements = []
        self.supports = []
        self.loads = {}         # node -> [fx, fy]
        self.reactions = None   # support forces, set by solve()

    def add_node(self, x, y):
        node = Node(x, y)
        node.index = len(self.nodes)
        self.nodes.append(node)
        return node

    def add_element(self, element):
        self.elements.append(element)
        return element

    def add_bar(self, node0, node1, E=210e9, A=2e-3):
        return self.add_element(Bar(node0, node1, E, A))

    def add_spring(self, node0, node1, k):
        return self.add_element(Spring(node0, node1, k))

    def add_support(self, support):
        self.supports.append(support)
        return support

    def add_load(self, node, fx, fy):
        self.loads[node] = [fx, fy]

    def clear_loads(self):
        self.loads = {}

    def solve(self):
        n_dofs = 2 * len(self.nodes)

        K = np.zeros((n_dofs, n_dofs))
        f = np.zeros(n_dofs)

        # Every element adds its own stiffness matrix. The loop doesn't know
        # or care whether an element is a Bar, a Spring or something else.

        for element in self.elements:
            K[np.ix_(element.dofs, element.dofs)] += element.stiffness_matrix()

        for node, (fx, fy) in self.loads.items():
            f[node.dofs] += [fx, fy]

        fixed = sorted({dof for support in self.supports for dof in support.fixed_dofs()})
        free = [dof for dof in range(n_dofs) if dof not in fixed]

        K_free = K[np.ix_(free, free)]

        if np.linalg.cond(K_free) > 1e12:
            raise ValueError("The structure is a mechanism, it can move without deforming. Add bars or supports.")

        u = np.zeros(n_dofs)
        u[free] = np.linalg.solve(K_free, f[free])

        for node in self.nodes:
            node.ux, node.uy = u[node.dofs]

        for element in self.elements:
            element.compute_force(u)

        self.reactions = K @ u - f

        return u

    def max_displacement(self):
        return max(np.hypot(node.ux, node.uy) for node in self.nodes)
````

### Building the bridge

A function in `bridge.py` uses the `Structure` interface to build a bridge with `n_panels` panels. The code reads almost like a description of the bridge: bottom chord, top chord, verticals, diagonals and bearings.

````python
def create_bridge(n_panels=6, panel_width=2.0, height=2.0, bearing_stiffness=5e7):
    """
    A Pratt truss bridge resting on two rubber bearings (springs).

    Returns the structure and the list of deck nodes (the bottom chord), where
    traffic loads are applied.
    """
    s = Structure()

    deck = [s.add_node(i * panel_width, 0.0) for i in range(n_panels + 1)]
    top = [None] + [s.add_node(i * panel_width, height) for i in range(1, n_panels)] + [None]

    for i in range(n_panels):
        s.add_bar(deck[i], deck[i + 1])                      # bottom chord

    for i in range(1, n_panels - 1):
        s.add_bar(top[i], top[i + 1])                        # top chord

    for i in range(1, n_panels):
        s.add_bar(deck[i], top[i])                           # verticals

    s.add_bar(deck[0], top[1])                               # end diagonals
    s.add_bar(top[n_panels - 1], deck[n_panels])

    for i in range(1, n_panels):                             # diagonals, sloping down towards the middle
        if i < n_panels / 2:
            s.add_bar(top[i], deck[i + 1])
        elif i > n_panels / 2:
            s.add_bar(top[i], deck[i - 1])

    # Bearings: a spring from each end of the bridge down to a fixed ground node.

    for node in [deck[0], deck[-1]]:
        ground = s.add_node(node.x, -1.0)
        s.add_support(PinSupport(ground))
        s.add_spring(node, ground, bearing_stiffness)

    # Something has to stop the bridge from sliding sideways.

    s.add_support(RollerSupport(deck[0], free="y"))

    return s, deck

def set_bearing_stiffness(structure, k):
    for element in structure.elements:
        if isinstance(element, Spring):
            element.k = k
````

### Analysis

Place a 100 kN load at the middle of the bridge, as in `01_bridge_plot.py`:

````python
LOAD = 100e3        # N, downwards
LOAD_NODE = 3       # index in the deck list, 3 = middle of the bridge

bridge, deck = create_bridge()
bridge.add_load(deck[LOAD_NODE], 0.0, -LOAD)
bridge.solve()

print("Element               Force [kN]")
for element in bridge.elements:
    print(f"{repr(element):20s} {round(element.force / 1e3, 1) + 0.0:10.1f}")  # + 0.0 turns -0.0 into 0.0

print()
print(f"Sum of vertical reactions: {bridge.reactions[1::2].sum() / 1e3:.1f} kN")
print(f"Max displacement:          {bridge.max_displacement() * 1e3:.2f} mm")
````

The sum of the vertical reactions equals the applied load, which is a good check of equilibrium. Next, plot the deformed bridge with the bars coloured by their axial force:

````python
from matplotlib.collections import LineCollection

# Displacements are a few mm, so scale them up to make them visible.

scale = 0.4 / bridge.max_displacement()

undeformed = []
deformed = []
forces = []
springs = []

for element in bridge.elements:
    n0, n1 = element.node0, element.node1

    undeformed.append([(n0.x, n0.y), (n1.x, n1.y)])
    line = [(n0.x + scale * n0.ux, n0.y + scale * n0.uy), (n1.x + scale * n1.ux, n1.y + scale * n1.uy)]

    if isinstance(element, Spring):
        springs.append(line)
    else:
        deformed.append(line)
        forces.append(element.force / 1e3)

fig, ax = plt.subplots(figsize=(11, 5))

ax.add_collection(LineCollection(undeformed, colors="lightgray", linewidths=1, linestyles="dashed"))
ax.add_collection(LineCollection(springs, colors="tab:green", linewidths=3))

limit = max(abs(force) for force in forces)
bars = LineCollection(deformed, cmap="coolwarm", linewidths=3)
bars.set_array(forces)
bars.set_clim(-limit, limit)
ax.add_collection(bars)
fig.colorbar(bars, ax=ax, orientation="horizontal", shrink=0.6, pad=0.15,
             label="Axial force [kN]  (blue = compression, red = tension)")

load_node = deck[LOAD_NODE]
x = load_node.x + scale * load_node.ux
y = load_node.y + scale * load_node.uy
ax.annotate(f"{LOAD / 1e3:.0f} kN", xy=(x, y), xytext=(x, 3.0), ha="center",
            arrowprops=dict(arrowstyle="->", linewidth=2))

ax.set_title(f"Pratt truss bridge, displacements scaled {scale:.0f}x")
ax.set_aspect("equal")
ax.autoscale()
ax.margins(0.08)
ax.set_ylim(-1.5, 3.5)      # room for the load arrow
ax.set_xlabel("x [m]")
ax.set_ylabel("y [m]")

fig.tight_layout()
plt.show()
````

Try moving the load by changing `LOAD_NODE`. You can also make the bearings softer with `set_bearing_stiffness(bridge, 1e7)` and solve again.

# 📚 Summary and Best Practices

## What We've Learned:

### 1. **Classes and Objects**
- Classes are blueprints, objects are instances
- `__init__()` method initializes objects
- `self` refers to the current instance

### 2. **Encapsulation**
- Use private attributes with `__` prefix
- Control access with properties (`@property`, `@setter`)
- Validate data in setter methods

### 3. **Inheritance**
- Child classes inherit from parent classes: `class Child(Parent):`
- Use `super()` to call parent methods
- Override methods to customize behavior

### 4. **Composition**
- Objects can contain other objects
- Use when you have "has-a" relationships
- More flexible than inheritance

### 5. **Polymorphism**
- Different objects can share the same interface
- Python uses duck typing
- Write code that works with multiple object types

### 6. **Special Methods**
- `__str__()` and `__repr__()` control how objects are printed
- `__add__()`, `__mul__()`, `__eq__()` etc. let objects use Python's operators

### 7. **Larger Programs**
- Put shared behaviour in base classes and let subclasses override what differs
- Use class attributes for values shared by all objects of a class
- Keep the model (simulation) separate from the view (graphics) in different modules

## Python OOP Best Practices:

### Class Design:
- **Use descriptive class names**: `BankAccount` not `BA`
- **Follow PascalCase**: `MyClass`, not `my_class`
- **Add docstrings**: Document what the class does
- **Keep classes focused**: Single responsibility principle

### Method Design:
- **Use snake_case for methods**: `calculate_area()`
- **Methods should do one thing**: Keep them simple
- **Use `@property` for computed values**: Don't store what you can calculate
- **Validate inputs**: Check types and ranges in setters

### Inheritance Guidelines:
- **Use inheritance for "is-a" relationships**: `Circle` is a `Shape`
- **Use composition for "has-a" relationships**: `Car` has an `Engine`
- **Prefer composition over inheritance**: More flexible
- **Keep inheritance hierarchies shallow**: Max 3-4 levels

### Code Organization:
- **Import at the top**: Keep imports organized
- **One class per file** (for larger projects)
- **Use meaningful variable names**: `radius` not `r`
- **Add type hints** (Python 3.5+): `def __init__(self, x: float, y: float)`

## 📋 Quick Reference Card

### Class Definition:
```python
class MyClass:
    """Class docstring"""
    
    def __init__(self, param):
        """Constructor"""
        self.__private_attr = param
    
    @property
    def public_attr(self):
        """Getter property"""
        return self.__private_attr
    
    @public_attr.setter  
    def public_attr(self, value):
        """Setter property with validation"""
        if value < 0:
            raise ValueError("Must be positive")
        self.__private_attr = value
    
    def public_method(self):
        """Public instance method"""
        return f"Value: {self.__private_attr}"
    
    def __str__(self):
        """String representation"""
        return f"MyClass({self.__private_attr})"
```

### Inheritance:
```python
class ChildClass(ParentClass):
    def __init__(self, child_param, parent_param):
        super().__init__(parent_param)  # Call parent constructor
        self.child_attr = child_param
    
    def overridden_method(self):
        """Override parent method"""
        result = super().overridden_method()  # Call parent method
        return f"Child: {result}"
```
